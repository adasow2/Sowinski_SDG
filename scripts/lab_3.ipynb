```python
import geopandas as gpd
import pandas as pd
from pathlib import Path
data_path = Path.cwd() / 'data'
```

```python
rivers = gpd.read_file(str(data_path / "hr1.geojson"))
demography = gpd.read_file(str(data_path / "lsp1.geojson"))
vulcans = gpd.read_file(str(data_path / "wulkany/lw1.geojson"))
```

```python
"""
zadanie 1
Dla warstwy z rzekami policz ile jest obiektów których
MultiLineString zawiera dokładnie między 2 a 10 części składowych geometrii
"""

rivers['l_geom'] = rivers.geometry.apply(lambda geom: len(geom.geoms))
rivers_filter = rivers[ rivers['l_geom'].between(2, 10)]
len(rivers_filter)
```

```python
"""
zadanie 2
Dla warstwy demograficznej wyświetl na mapie kraje,
w których odsetek osób w wieku 65+ wynosi ponad 18%
"""

demography['65+'] = (
    demography['ludność_w_wieku_65_lat_i_więcej'].astype(str)
    .str.replace("%", "", regex=False)
    .str.replace(r"\s+", "", regex=True)
)

demography['65+'] = pd.to_numeric(demography['65+'], errors='coerce')
demography.dropna(subset=['65+'], inplace=True)
demography_filter = demography[ demography['65+'] > 18 ]
demography_filter.plot()
```
